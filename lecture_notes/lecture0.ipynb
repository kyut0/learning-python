## Strings (str)

Collection of characters strung together, stored inside quotation marks (single or double, must be consistent)

### fstrings
Used to inject a variable into a string.  
To use this technique: f"Some string literal, plus some {variable}"

````python
name = input("What is your name? ")

print(f"Hello, {name}!")
````

### String cleanup

We can clean up strings using functions that are built into python: 
- .strip() to remove whitespace
- .capitalize() to capitalize the first letter of the string
- .title() to capitalize the first letter of each word in the string

````python
# Ask user for their name
name = input("What is your name? ")

# Remove whitespace from the left and right of name
name = name.strip() # Converts "     david    " to "david"

# Fix capitalization of name
name = name.capitalize() # Converts "david" to "David"

# Fix capitalization of last name
name = name.title() # Converts "david smith" to "David Smith"

# Say hello to user
print(f"Hello, {name}!")
````

**NOTE**:

We don't need to use all these lines of code--we can simplify/collapse these function calls by chaining them together.

HOWEVER, note that this is not necessary, and both versions of the code are functionally identical. Personal preference. 

````python
# Ask the user for their name -- AND remove whitespace and fix capitalization
name = input("What is your name? ").strip().title()

# Say hello to user
print(f"Hello, {name}!")
````

More string methods:

````python
# Ask user for their name
name = input("What is your name? ").strip().title()

# Split name into first and last name
first, last = name.split(" ") # Splits "David Smith" into ["David", "Smith"]

# Say hello to user (by just first name)
print(f"Hello, {first}!")
````

## Integers (int)

Numbers, numeric values, etc. Can be used in mathematical functions. (+, -, /, *. %)

````python
x = input("What is x? ") # type 1
y = input("What is y? ") # type 2

z = x + y

print(z)
````

Oh no! What happened? x and y are strings, and by adding them we simply concatenated the strings (rather than adding their values).

To fix:

````python
x = input("What is x? ") # type 1
y = input("What is y? ") # type 2

z = int(x) + int(y)

print(z)
````

````python
# Even better solution:
x = int(input("What is x? ")) # type 1
y = int(input("What is y? ")) # type 2

print(x + y)
````

````python
# One line solution:
print(int(input("What is x? ")) + int(input("What is y? ")))
````

Is the shortest version the best? No, probably not. Getting too nested and too complex.

## Float

Numbers with decimal values

````python
# Even better solution:
x = float(input("What is x? ")) # type 1.2
y = float(input("What is y? ")) # type 4.2

print(x + y)
````

### round()

````python
# Even better solution:
x = float(input("What is x? ")) # type 1.2
y = float(input("What is y? ")) # type 4.2

z = round(x + y)

print(z)
````

````python
# Even better solution:
x = float(input("What is x? ")) # type 1.2
y = float(input("What is y? ")) # type 4.2

z = round(x + y)

print(f"{z}")
````
